### Programming problem 1
Consider the problem to compute the product of two positive integers, $n$ and $m$, using only addition and subtraction.
This programming problem is to consider the recursive approach in above problem and asks you to complete the following tasks.
1. Implement the approach as a function named as `product_rec` using Python.
2. Please have an iterative version for the approach and write a function for this version with function name `product_ite`.
3. Compare these two function with the same input in terms of running time and write what you have observed.

**Note** that we will use an in-built python library **`timeit`** and the module function **`timeit.timeit()`** for measuring the running time.
For the programming problems, please do some experiments by yourself to observe the running time and the number of recursive calls. You need to report what you have observed from the experiments when you submit your homework.

```python
def product_rec(n, m):
    if n == 1 :
        return m
    else:
        return n + product_rec(n - 1, m)
print("The product of 4 and 7 is", product_rec(4, 7))
```

```python
def product_ite(n, m):
    result = 0
    for _ in range(n):
        result += m
    return result

print("The product of 4 and 7 is", product_ite(4, 7))
```

```python
#---------------------------------------------------------timeit
import timeit as ti

#---------------------------------------------------------Method 1
print("Running time comparison (Timer Method 1):")
tstart = ti.default_timer() #timer start 計時器開啟
for i in range(10000):
    product_rec(40, 700)
tend = ti.default_timer() #timer end計時器關閉
t_rec=tend - tstart#開始與結束的差值

tstart = ti.default_timer() #timer start 計時器開啟
for i in range(10000):
    product_ite(40, 700)
tend = ti.default_timer() #timer end計時器關閉
t_ite=tend - tstart#開始與結束的差值
print("Recursion 40*700=", product_rec(40, 700),"Iteration 40*700=", product_ite(40, 700))
print("Recursive approach:", t_rec)
print("iterative approach:", t_ite)

#---------------------------------------------------------Method 2
print("Running time comparison (Timer Method 2):")
t_rec = ti.timeit("product_rec(40, 700)", setup="from __main__ import product_rec", number = 10000)#timeit(函數名稱,執行次數)
t_ite = ti.timeit("product_ite(40, 700)", setup="from __main__ import product_ite", number = 10000)

print("Recursion 40*700=", product_rec(40, 700),"Iteration 40*700=", product_ite(40, 700))
print("Recursive approach:", t_rec)
print("Iterative approach:", t_ite)
```

```python
def chebyshev_recursive(n):
    """Return coefficients of Chebyshev polynomial T_n(x) (first kind) recursively."""
    if n == 0:
        return [1]        # T0(x) = 1
    elif n == 1:
        return [1, 0]     # T1(x) = x

    Tn_1 = chebyshev_recursive(n - 1)
    Tn_2 = chebyshev_recursive(n - 2)

    Tn_1_shifted = [2 * c for c in Tn_1] + [0]


    len_diff = len(Tn_1_shifted) - len(Tn_2)
    if len_diff > 0:
        Tn_2 = [0] * len_diff + Tn_2
    elif len_diff < 0:
        Tn_1_shifted = [0] * (-len_diff) + Tn_1_shifted


    Tn = [a - b for a, b in zip(Tn_1_shifted, Tn_2)]
    return Tn  


if __name__ == "__main__":
    for n in range(11):
        print(f"The coefficients of T_{n}(x) in list = {chebyshev_recursive(n)}")
```

```python
def chebyshev_memoreccu(n, memo):
    if memo[n] is not None:
        return memo[n]

    if n == 0:
        memo[0] = [1]        # T0(x) = 1
        return memo[0]
    if n == 1:
        memo[1] = [1, 0]     # T1(x) = x
        return memo[1]

    # 遞迴（使用快取）
    Tn_1 = chebyshev_memoreccu(n - 1, memo)  # T_{n-1}
    Tn_2 = chebyshev_memoreccu(n - 2, memo)  # T_{n-2}

    two_x_Tn1 = [c * 2 for c in Tn_1] + [0]

    L = max(len(two_x_Tn1), len(Tn_2))
    a = [0] * (L - len(two_x_Tn1)) + two_x_Tn1
    b = [0] * (L - len(Tn_2))     + Tn_2

    Tn = [ai - bi for ai, bi in zip(a, b)]
    memo[n] = Tn
    return Tn


if __name__ == "__main__":
    for n in range(11):
        m_table = [None] * (n + 1)    
        coeffs = chebyshev_memoreccu(n, m_table)
        print(f"The coefficients of T_{n}(x) in list = {coeffs}")
```

```python
def chebyshev_addoutrec(n, Tn_1=None, Tn_2=None, k=1):
    if n == 0:
        return [1]        # T0(x) = 1
    elif n == 1:
        return [1, 0]     # T1(x) = x

    if Tn_1 is None and Tn_2 is None:
        Tn_2 = [1]        # T0
        Tn_1 = [1, 0]     # T1
        k = 1

    if k == n:
        return Tn_1

    two_x_Tn1 = [c * 2 for c in Tn_1] + [0]

    L = max(len(two_x_Tn1), len(Tn_2))
    a = [0] * (L - len(two_x_Tn1)) + two_x_Tn1
    b = [0] * (L - len(Tn_2)) + Tn_2

    T_next = [ai - bi for ai, bi in zip(a, b)]
    return chebyshev_addoutrec(n, T_next, Tn_1, k + 1)


if __name__ == "__main__":
    for n in range(11):
        print(f"The coefficients of T_{n}(x) in list = {chebyshev_addoutrec(n)}")
```

```python
def chebyshev_iterative(n):
    
    if n == 0:
        return [1]          # T0(x) = 1
    elif n == 1:
        return [1, 0]       # T1(x) = x

    Tn_2 = [1]              # T0
    Tn_1 = [1, 0]           # T1

    for k in range(2, n + 1):
        two_x_Tn1 = [c * 2 for c in Tn_1] + [0]

        L = max(len(two_x_Tn1), len(Tn_2))
        a = [0] * (L - len(two_x_Tn1)) + two_x_Tn1
        b = [0] * (L - len(Tn_2)) + Tn_2

        Tn = [ai - bi for ai, bi in zip(a, b)]

        Tn_2, Tn_1 = Tn_1, Tn

    return Tn_1 


if __name__ == "__main__":
    for n in range(11):
        print(f"The coefficients of T_{n}(x) in list = {chebyshev_iterative(n)}")
```
