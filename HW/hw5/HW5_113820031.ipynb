```python
import re

# Node class
class Hnode:
    def __init__(self, key, item):
        self.key = key
        self.item = item
        self.parent = None
        self.right = None
        self.left = None

    def getKey(self):
        return self.key

    def getItem(self):
        return self.item

    def getRightChild(self):
        return self.right

    def getLeftChild(self):
        return self.left

    def getParent(self):
        return self.parent

    def hasRightChild(self):
        return (self.right != None)

    def hasLeftChild(self):
        return (self.left != None)

    def isRoot(self):
        return (self.parent == None)

    def isLeaf(self):
        return ((self.right == None) and (self.left == None))

    def isRightChild(self):
        return (self.parent.right == self)

    def isLeftChild(self):
        return (self.parent.left == self)

    def setParent(self, p):
        self.parent = p

    def setKey(self, key):
        self.key = key

    def setItem(self, item):
        self.item = item

    def addRightChild(self, hnode):
        self.right = hnode
        hnode.setParent(self)

    def addLeftChild(self, hnode):
        self.left = hnode
        hnode.setParent(self)


# Heap class
class Heap:
    def __init__(self):
        self.root = None
        self.last = None
        self.size = 0

    def isEmpty(self):
        return (self.size == 0)

    def getSize(self):
        return self.size

    def getRoot(self):
        return self.root

    def getLast(self):
        return self.last

    def getHighestPriority(self):
        return self.root.getKey()

    def setRoot(self, hnode):
        self.root = hnode

    def setLast(self, hnode):
        self.last = hnode


    def _nodeAt(self, idx: int):
        if idx <= 0 or self.root is None:
            return None
        if idx == 1:
            return self.root

        bits = bin(idx)[3:] 
        cur = self.root
        for b in bits:
            if cur is None:
                return None
            if b == '0':
                cur = cur.getLeftChild()
            else:
                cur = cur.getRightChild()
        return cur

    def _swapEntry(self, a: Hnode, b: Hnode):
        a.key, b.key = b.key, a.key
        a.item, b.item = b.item, a.item

 
    def downwardHeapify(self, current):
        if current is None:
            return

        while True:
            smallest = current

            if current.hasLeftChild() and current.getLeftChild().getKey() < smallest.getKey():
                smallest = current.getLeftChild()

            if current.hasRightChild() and current.getRightChild().getKey() < smallest.getKey():
                smallest = current.getRightChild()

            if smallest is current:
                break

            self._swapEntry(current, smallest)
            current = smallest

    def upwardHeapify(self, current):
        if current is None:
            return

        while (not current.isRoot()) and current.getKey() < current.getParent().getKey():
            self._swapEntry(current, current.getParent())
            current = current.getParent()


    def removeMin(self):
        if self.isEmpty():
            print("The heap is empty and no entry can be removed")
            return None


        if self.size == 1:
            removed_key = self.root.getKey()
            removed_item = self.root.getItem()
            self.root = None
            self.last = None
            self.size = 0
            print("The heap is now empty")
            return (removed_key, removed_item)

        removed_key = self.root.getKey()
        removed_item = self.root.getItem()

        last_node = self.last
        self.root.setKey(last_node.getKey())
        self.root.setItem(last_node.getItem())

        p = last_node.getParent()
        if p is not None:
            if p.getLeftChild() == last_node:
                p.left = None
            elif p.getRightChild() == last_node:
                p.right = None

        self.size -= 1
        self.last = self._nodeAt(self.size)

        self.downwardHeapify(self.root)

        if self.size == 0:
            print("The heap is now empty")

        return (removed_key, removed_item)


    def Insert(self, hnode):
        if hnode is None:
            return

        if self.isEmpty():
            self.root = hnode
            self.last = hnode
            self.size = 1
            return

        new_idx = self.size + 1
        parent_idx = new_idx // 2
        parent = self._nodeAt(parent_idx)

        if parent is None:
            raise RuntimeError("Heap structure corrupted: cannot find parent.")

        if new_idx % 2 == 0:
            parent.addLeftChild(hnode)
        else:
            parent.addRightChild(hnode)

        self.size += 1
        self.last = hnode

        # 向上調整
        self.upwardHeapify(hnode)

    #
    # print the heap in pre-order
    #
    def printHeapPreOrder(self, i):
        if i is None:
            return

        if i.isLeaf():
            print(f"Leaf [ {i.getKey()} {i.getItem()} ]")
            return
        else:
            print(f"Node [ {i.getKey()} {i.getItem()} ]")

        if i.hasLeftChild():
            self.printHeapPreOrder(i.getLeftChild())
        if i.hasRightChild():
            self.printHeapPreOrder(i.getRightChild())


# decimal to binary using yield (保留你原本的)
def d2by(x):
    while x > 0:
        x, r = divmod(x, 2)
        yield r


# function for starting the task
def HeapwithEntriesInserted():
    #
    # read the input information from the default input text file into an
    # entry list, entry_list
    #
    entry_list = []
    try:
        with open("inFile.txt", "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                parts = line.split()
                key = int(parts[0])
                item = " ".join(parts[1:]) if len(parts) > 1 else ""
                entry_list.append((key, item))
    except FileNotFoundError:
        print("找不到 inFile.txt，請確認檔案在同一個資料夾。")
        return

    #
    # initiating a heap object h
    #
    h = Heap()

    #
    # Do something here to build the heap
    #
    for key, item in entry_list:
        h.Insert(Hnode(key, item))

    #---------------------Print as the example on the homework sheet --------
    print("Heap size=", h.getSize(), "The highest priority is ", h.getHighestPriority())
    print("pre-order traversal:")
    h.printHeapPreOrder(h.getRoot())

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("deleteMin")
    h.removeMin()

    print("insert 35, resume")
    h.Insert(Hnode(35, "resume"))

    print("insert 15, second")
    h.Insert(Hnode(15, "second"))

    print("insert 20, fourth")
    h.Insert(Hnode(20, "fourth"))

    print("Heap size=", h.getSize(), "The highest priority is ", h.getHighestPriority())
    print("pre-order traversal:")
    h.printHeapPreOrder(h.getRoot())

    print("deleteMin")
    h.removeMin()

    print("insert 40, nineth")
    h.Insert(Hnode(40, "nineth"))


HeapwithEntriesInserted()
```
